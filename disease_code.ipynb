```python
df.head()
```

```python
df['time.ppn']= df['time.ppn'].fillna(df['time.ppn'].median())
```

```python
df['fasting']= (df['time.ppn'] >= 480).astype(int)
```

```python
df = df.drop(['id','location','bp.2s','bp.2d','time.ppn'],axis = 1)
```

```python
df.head()
```

```python

df['height'] = df['height'].fillna(df['height'].median())
df['weight'] = df['weight'].fillna(df['weight'].median())
df['hdl'] = df['hdl'].fillna(df['hdl'].median())
df['waist'] = df['waist'].fillna(df['waist'].median())
df['hip'] = df['hip'].fillna(df['hip'].median())
df['ratio'] = df['ratio'].fillna(df['ratio'].median())
df['chol'] = df['chol'].fillna(df['chol'].median())
df['bp.1s'] = df['bp.1s'].fillna(df['bp.1s'].median())
df['bp.1d'] = df['bp.1d'].fillna(df['bp.1d'].median())
df['frame'] = df['frame'].fillna(df['frame'].mode()[0])
```

```python
df.isnull().sum()
```

```python
df = df.dropna(subset=['glyhb'])
```

```python
df['diabetic'] = (df['glyhb'] >= 5.7).astype(int)
```

```python
df['gender'] = (df['gender'] == 'male').astype(int)
```

```python
df = pd.get_dummies(df, columns=['frame'], drop_first=True, dtype=int)
```

```python
df.head()
```

```python
df['whr'] = df['waist'] / df['hip']
df['glucose_fasting_interaction'] = df['stab.glu'] * df['fasting']
df['chol_ratio_interaction'] = df['chol'] * df['ratio']
df['age_waist_interaction'] = df['age'] * df['waist']
df['bmi'] = (df['weight'] / (df['height'] ** 2)) * 703
```

```python
X_d = df.drop(columns=['diabetic', 'glyhb'])
y_d = df['diabetic']
```

```python
from sklearn.feature_selection import mutual_info_classif
mi_scores = mutual_info_classif(X_d, y_d)
mi_df = pd.Series(mi_scores, index=X_d.columns).sort_values(ascending=False)
print(mi_df)
```

```python
import re

df_k = df_k.iloc[2:].reset_index(drop=True)  # drop metadata rows

def convert_range(val):
    if pd.isna(val):
        return None
    val = str(val).strip()
    if val.startswith('≥') or val.startswith('>='):
        return float(re.sub(r'[≥>=]', '', val).strip())
    if val.startswith('<'):
        return float(val.replace('<', '').strip())
    if '-' in val:
        parts = val.split('-')
        try:
            return (float(parts[0].strip()) + float(parts[1].strip())) / 2
        except:
            return None
    try:
        return float(val)
    except:
        return None

range_cols = ['sg','al','su','bgr','bu','sod','sc','pot','hemo','pcv','rbcc','wbcc','grf','age']
for col in range_cols:
    df_k[col] = df_k[col].apply(convert_range)

# target
df_k['diagnosis'] = (df_k['class'].str.strip() == 'ckd').astype(int)

# drop leakage-risk columns
df_k = df_k.drop(columns=['class', 'grf', 'stage', 'affected'])

# convert remaining binary columns to int
binary_cols = ['bp (Diastolic)','bp limit','rbc','pc','pcc','ba','htn','dm','cad','appet','pe','ane']
for col in binary_cols:
    df_k[col] = df_k[col].astype(int)

# sanity checks
print(df_k.shape)
print(df_k.isna().sum().sum(), 'missing values')
print(df_k['diagnosis'].value_counts())
print(df_k.dtypes)
```

## diabetes rf 

```python
X_train_d,X_test_d,y_train_d,y_test_d = train_test_split(X_d,y_d,test_size=0.3,stratify=y_d,random_state = 42)
```

```python
rf = RandomForestClassifier(class_weight='balanced',random_state=42)
params = {'n_estimators':[100,150,200,250,300],'criterion':['gini', 'entropy', 'log_loss'],'ccp_alpha':[0.0, 0.001, 0.005, 0.01, 0.02]}
grid_d = GridSearchCV(estimator=rf,param_grid=params,scoring='recall',cv=5)
```

```python
grid_d.best_params_
```

```python
print(accuracy_score(y_test_d, y_pred_d))
print(confusion_matrix(y_test_d, y_pred_d))   
print(classification_report(y_test_d, y_pred_d))
```

```python
from sklearn.metrics import precision_recall_curve
import pandas as pd

rf_probs_d = grid_d.predict_proba(X_test_d)[:, 1]  # unscaled, matching how you trained it

print("RF probs range:", rf_probs_d.min(), rf_probs_d.max())  # sanity check first

precisions, recalls, thresholds = precision_recall_curve(y_test_d, rf_probs_d)

df_thresh_rf_d = pd.DataFrame({
    'threshold': thresholds,
    'precision': precisions[:-1],
    'recall': recalls[:-1]
})

print(df_thresh_rf_d[df_thresh_rf_d['precision'] >= 0.70].sort_values('recall', ascending=False).head(20))
```

```python
from sklearn.model_selection import cross_val_score
cv_scores_rf_d = cross_val_score(grid_d.best_estimator_, X_train_d, y_train_d, cv=5, scoring='recall')
print(cv_scores_rf_d.mean(), cv_scores_rf_d.std())
```

```python


def load_disease_pipeline(disease_name):
    """Loads all models, scaler, and thresholds for a given disease."""
    models = {
        'rf': joblib.load(f'{disease_name}_rf.pkl'),
        'knn': joblib.load(f'{disease_name}_knn.pkl'),
        'svm': joblib.load(f'{disease_name}_svm.pkl'),
        'nb': joblib.load(f'{disease_name}_nb.pkl'),
    }
    scaler = joblib.load(f'{disease_name}_scaler.pkl')

    with open(f'{disease_name}_thresholds.json', 'r') as f:
        thresholds = json.load(f)

    return models, scaler, thresholds


def predict_pipeline(patient_data_raw, patient_data_scaled, models, thresholds):
    """
    patient_data_raw: unscaled features (for RF, NB) - shape (1, n_features)
    patient_data_scaled: scaled features (for KNN, SVM) - shape (1, n_features)
    """
    # Stage 1 — RF (unscaled)
    rf_proba = models['rf'].predict_proba(patient_data_raw)[0][1]

    if rf_proba >= thresholds['rf']:
        return {
            'prediction': 1,
            'confidence': rf_proba,
            'stage_used': 'Stage 1 (Random Forest)'
        }

    # Stage 2 — KNN + SVM (scaled), NB (unscaled)
    knn_proba = models['knn'].predict_proba(patient_data_scaled)[0][1]
    svm_proba = models['svm'].predict_proba(patient_data_scaled)[0][1]
    nb_proba = models['nb'].predict_proba(patient_data_raw)[0][1]

    knn_pred = int(knn_proba >= thresholds['knn'])
    svm_pred = int(svm_proba >= thresholds['svm'])
    nb_pred = int(nb_proba >= thresholds['nb'])

    votes = knn_pred + svm_pred + nb_pred
    final_pred = 1 if votes >= 2 else 0  # majority vote

    avg_confidence = (knn_proba + svm_proba + nb_proba) / 3

    return {
        'prediction': final_pred,
        'confidence': avg_confidence,
        'stage_used': 'Stage 2 (Ensemble - Majority Vote)',
        'individual_votes': {'knn': knn_pred, 'svm': svm_pred, 'nb': nb_pred}
    }
```

```python
heart_normal_ranges = {
    'trestbps': (90, 130),      # resting blood pressure (mm Hg)
    'chol': (125, 200),         # serum cholesterol (mg/dl)
    'thalach': (100, 170),      # max heart rate achieved — context-dependent on age, rough general band
    'oldpeak': (0, 1),          # ST depression — above 1 suggests more cardiac stress
    'ca': (0, 0),               # number of major vessels blocked — 0 is ideal, >0 flags concern
    'age': (0, 55),             # flags "older age" as a risk contributor beyond this
}

heart_clinical_advice = {
    'trestbps': "your resting blood pressure is elevated — consider reducing sodium intake and increasing physical activity",
    'chol': "your cholesterol is above the healthy range — consider dietary changes and discussing statin therapy with your doctor",
    'thalach': "your maximum heart rate response is lower than typical — this may reflect reduced cardiovascular fitness, worth discussing with your doctor",
    'oldpeak': "your ST depression value suggests notable cardiac stress during exercise — this warrants a cardiology follow-up",
    'ca': "you have blocked major vessels detected via imaging — this is a significant finding requiring cardiology follow-up",
    'age': "age is a non-modifiable risk factor, but managing other risk factors becomes more important as you get older",
}

def generate_heart_explanation(patient_data, rf_model, feature_names):
    importances = pd.Series(rf_model.feature_importances_, index=feature_names).sort_values(ascending=False)

    reasons = []
    for feature in importances.index:
        if feature not in heart_normal_ranges:
            continue
        value = patient_data[feature].values[0]
        low, high = heart_normal_ranges[feature]
        if value < low or value > high:
            advice = heart_clinical_advice.get(feature, "consider discussing this with your doctor")
            reasons.append(f"{feature} ({value:.1f}) is outside the typical range — {advice}")
        if len(reasons) >= 3:
            break

    return reasons if reasons else ["No individually abnormal high-impact factors found, but overall pattern still suggests risk — consult a doctor."]
```

```python
ckd_normal_ranges = {
    'bgr': (70, 140),      # blood glucose random (mg/dl)
    'bu': (7, 20),         # blood urea (mg/dl)
    'sc': (0.6, 1.3),      # serum creatinine (mg/dl) — key kidney marker
    'sod': (135, 145),     # sodium (mEq/L)
    'pot': (3.5, 5.0),     # potassium (mEq/L)
    'hemo': (12, 17),      # hemoglobin (g/dl)
    'pcv': (36, 50),       # packed cell volume (%)
    'sg': (1.015, 1.025),  # urine specific gravity
    'al': (0, 0),          # albumin in urine — 0 is normal, any presence flags kidney concern
}

ckd_clinical_advice = {
    'bgr': "your blood glucose is outside the normal range — managing blood sugar can help protect kidney function",
    'bu': "your blood urea level is abnormal — this reflects how well your kidneys are filtering waste, worth discussing with a nephrologist",
    'sc': "your serum creatinine is outside the normal range — this is a key marker of kidney filtration function and needs medical follow-up",
    'sod': "your sodium level is abnormal — this can affect fluid balance and kidney function, worth monitoring with your doctor",
    'pot': "your potassium level is outside the normal range — this is important to monitor closely, especially with kidney concerns",
    'hemo': "your hemoglobin is low — reduced kidney function can affect red blood cell production, discuss with your doctor",
    'pcv': "your packed cell volume is abnormal — this may relate to anemia associated with kidney function decline",
    'sg': "your urine specific gravity is abnormal — this reflects how well your kidneys are concentrating urine",
    'al': "protein (albumin) detected in your urine — this is an important early indicator of kidney damage and should be followed up",
}

def generate_ckd_explanation(patient_data, rf_model, feature_names):
    importances = pd.Series(rf_model.feature_importances_, index=feature_names).sort_values(ascending=False)

    reasons = []
    for feature in importances.index:
        if feature not in ckd_normal_ranges:
            continue
        value = patient_data[feature].values[0]
        low, high = ckd_normal_ranges[feature]
        if value < low or value > high:
            advice = ckd_clinical_advice.get(feature, "consider discussing this with your doctor")
            reasons.append(f"{feature} ({value:.2f}) is outside the typical range — {advice}")
        if len(reasons) >= 3:
            break

    return reasons if reasons else ["No individually abnormal high-impact factors found, but overall pattern still suggests risk — consult a doctor."]
```

```python
diabetes_normal_ranges = {
    'stab.glu': (70, 140),                    # stabilized glucose (mg/dl)
    'chol': (125, 200),                       # total cholesterol (mg/dl)
    'ratio': (0, 5),                          # cholesterol/HDL ratio — above 5 is a risk indicator
    'hdl': (40, 100),                         # HDL cholesterol (mg/dl) — below 40 is a risk factor
    'bp.1s': (90, 130),                       # systolic blood pressure
    'bp.1d': (60, 85),                        # diastolic blood pressure
    'waist': (0, 40),                         # waist circumference (inches) — general risk threshold
    'whr': (0, 0.9),                          # waist-to-hip ratio — above 0.9 (men) / 0.85 (women) is a risk marker
    'age_waist_interaction': (0, 2000),       # engineered feature — flag high compounded risk
    'glucose_fasting_interaction': (0, 140),  # engineered feature — flags high fasting glucose specifically
}

diabetes_clinical_advice = {
    'stab.glu': "your glucose level is elevated — reducing sugar and refined carbohydrate intake can help lower this",
    'chol': "your total cholesterol is above the healthy range — consider dietary changes and discussing this with your doctor",
    'ratio': "your cholesterol-to-HDL ratio is elevated, suggesting higher cardiovascular/metabolic risk — regular exercise can help improve this",
    'hdl': "your HDL ('good') cholesterol is lower than ideal — regular aerobic exercise can help raise it",
    'bp.1s': "your systolic blood pressure is elevated — reducing sodium intake and increasing physical activity may help",
    'bp.1d': "your diastolic blood pressure is elevated — lifestyle changes like reducing salt and stress can help",
    'waist': "your waist circumference suggests higher abdominal fat, a known diabetes risk factor — regular exercise and dietary changes can help",
    'whr': "your waist-to-hip ratio indicates higher abdominal fat distribution, which is strongly linked to insulin resistance",
    'age_waist_interaction': "the combination of your age and waist measurement suggests compounded metabolic risk — worth discussing with your doctor",
    'glucose_fasting_interaction': "your fasting glucose reading is elevated — this is a key diabetes screening indicator and should be discussed with your doctor",
}

def generate_diabetes_explanation(patient_data, rf_model, feature_names):
    importances = pd.Series(rf_model.feature_importances_, index=feature_names).sort_values(ascending=False)

    reasons = []
    for feature in importances.index:
        if feature not in diabetes_normal_ranges:
            continue
        value = patient_data[feature].values[0]
        low, high = diabetes_normal_ranges[feature]
        if value < low or value > high:
            advice = diabetes_clinical_advice.get(feature, "consider discussing this with your doctor")
            reasons.append(f"{feature} ({value:.2f}) is outside the typical range — {advice}")
        if len(reasons) >= 3:
            break

    return reasons if reasons else ["No individually abnormal high-impact factors found, but overall pattern still suggests risk — consult a doctor."]
```
